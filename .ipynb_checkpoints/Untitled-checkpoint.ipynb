```python
# The goal here is to formulate FBA as an optimization problem which we can solve
# via gradient descent.

# Specifically, we have the objective max_v{cv} subject to Sv = 0
# cv = biomass/growth rate, so we are aiming to find the fluxes which maximize growth
# subject to the constraint Sv = 0 (which means subject to them being viable fluxes)
# c is therefore a one-hot vector with the same dimension as V, which is all zeros except
# for the biomass reaction, where it is 1. This is how cV gives us the biomass reaction flux.

# We will try to predict the optimal vector v given an (N, 2) vector of 
# lower and upper bounds per reaction.
# for this initial experiment, we will use a simple MLP, and we will randomly
# knock out 1-3 genes, simulate the resulting fluxes with model.optimize()


# Implementation notes
"""
* We will use reaction bounds as inputs rather than genes. Modifying the reaction bounds simulates dropping the genes. Gene interactions
with the reactions are non-trivial - a gene can affect multiple reactions, and has complex interactions with enzymes and proteins. 

* We will handle scaling of outputs by normalizing them relative to the mean and std dev of the reaction's flux distribution. These
will be approximated by monte carlo sampling per reaction. This choice is because acceptable flux ranges for each reaction vary wildly,
and are often long-tailed distributions. We could normalize to [0, 1] or [-1, 1] with an activation function, but would not match the flux
distribution AND would then need to find an appropriate scale factor anyway.

* After selecting gene knockouts, and obtaining the GT flux vector, we must check if the solution is feasible. We can do this
by insepcting the '.status' of model.optimize() and keeping only examples which have status 'optimal'

* The flux vector is large but may be sparse. We can identify, during monte carlo sampling, reactions with 0 variance, and drop these
from our simulation.
"""
```

```python
import cobra
from cobra.io import load_model
# This is an E.Coli model that comes bundled with cobrapy
model = load_model("iJO1366")
# Do this just once at the beginning
model, removed_rxns = cobra.manipulation.prune_unused_reactions(model)
```

```python
import numpy as np
means = np.zeros_like(model.reactions)
sq_difs_from_mean = np.zeros_like(model.reactions)
stds = np.zeros_like(model.reactions)
```

```python
count = 0

# Welford's algorithm for maintaining an online mean and std
for i in range(50000):
    num_genes = np.random.randint(low=1, high=3)
    #print(num_genes)
    gene_idxs = np.random.randint(low=0, high=len(model.genes), size=(num_genes,))
    #print(gene_idxs, type(gene_idxs), gene_idxs.dtype)
    drop_genes = list(np.take(model.genes, gene_idxs))
    #print(drop_genes)
    with model:
        # Using delete_model_genes will modify the reaction network, which we cannot have
        # because we need the reactions/fluxes to stay fixed. We assume a static graph.
        cobra.manipulation.knock_out_model_genes(model, drop_genes)
        # Check validity
        solution = model.optimize()
        if solution.status == 'optimal':
            count += 1
            delta = solution.fluxes - means
            means = means + delta/count
            sq_difs_from_mean = sq_difs_from_mean + delta*(solution.fluxes - means)
    if i % 1000 == 0:
        print(i)

variances = sq_difs_from_mean / (count - 1)
stds = np.pow(variances, 0.5)
```
